```python
def get_bot_response(message, history):

    msg = message.lower().strip()

    if any(w in msg for w in ["hello", "hi", "hey", "namaste"]):
        return "Hello! Welcome to Customer Support. How can I help you?"

    elif "your name" in msg:
        return "I am HelpBot, your customer service assistant."

    elif "working hours" in msg:
        return "Our working hours are Monday to Saturday, 9 AM to 6 PM."

    elif "contact" in msg or "email" in msg:
        return "You can contact us at support@example.com."

    elif "order" in msg or "delivery" in msg or "track" in msg:
        return "Please share your order ID with our support team."

    elif "return" in msg or "refund" in msg:
        return "For returns and refunds, contact our support team."

    elif "payment" in msg:
        return "For payment issues, please contact customer support."

    elif "thank" in msg:
        return "You're welcome! Have a nice day."

    elif "bye" in msg:
        return "Thank you for contacting us!"

    else:
        return "Sorry, I don't understand. Please ask another question."
```

```python
import gradio as gr

demo = gr.ChatInterface(
    fn=get_bot_response,
    title="🤖 HelpBot - Customer Service",
    description="Welcome! Ask me anything about orders, payments, delivery and support.",
    examples=[
        "Hello",
        "What are your working hours?",
        "How can I track my order?",
        "How can I contact support?",
        "How can I get a refund?"
    ]
)

demo.launch(share=True)
```
